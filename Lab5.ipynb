```python
import pandas as pd
from scipy.stats import zscore

# 1. Load the dataset
df = pd.read_csv("AAPL.csv")
print("Original Dataset:")
display(df.head())

# 2. Select numerical columns
numeric_cols = df.select_dtypes(include="number").columns

# 3. Detect outliers using IQR
print("\n--- IQR Outlier Detection ---")

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: {count} outliers")

# 4. Detect outliers using Z-score
print("\n--- Z-Score Outlier Detection ---")

for col in numeric_cols:
    values = df[col]
    scores = pd.Series(float("nan"), index=df.index)
    valid = values.notna()

    if values[valid].nunique() > 1:
        scores.loc[valid] = zscore(values[valid])

    count = (scores.abs() > 3).sum()
    print(f"{col}: {count} outliers")

# 5. Apply IQR-based capping
cleaned_df = df.copy()

print("\n--- Applying Capping ---")

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    cleaned_df[col] = df[col].clip(lower=lower, upper=upper)
    print(f"{col}: Capping completed")

# 6. Display cleaned dataset
print("\n--- Cleaned Dataset ---")
display(cleaned_df.head())

# 7. Save cleaned dataset
cleaned_df.to_csv("AAPL_cleaned.csv", index=False)

print("\nCleaned dataset saved as AAPL_cleaned.csv")
print("Original shape:", df.shape)
print("Cleaned shape:", cleaned_df.shape)
```

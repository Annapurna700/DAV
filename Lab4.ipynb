```python


import pandas as pd
import numpy as np

# Create a sample dataset
data = {
    'Name': ['Asha', 'Ravi', 'Priya', 'Kiran', 'Rahul'],
    'Age': [20, np.nan, 22, 25, np.nan],
    'Salary': [20000, 25000, np.nan, 30000, 35000],
    'Gender': ['Female', 'Male', 'Female', 'Male', 'Male'],
    'City': ['Mysuru', 'Bengaluru', 'Mysuru', np.nan, 'Hassan']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# 1. Identify missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 2. Handle missing values using Mean Imputation
df['Age'] = df['Age'].fillna(df['Age'].mean())
print("\nAfter Mean Imputation:")
print(df)

# 3. Handle missing values using Median Imputation
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
print("\nAfter Median Imputation:")
print(df)

# 4. Handle missing values using Forward Fill
df['City'] = df['City'].ffill()
print("\nAfter Forward Fill:")
print(df)

# 5. Encode categorical variables using Label Encoding
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Gender_Encoded'] = le.fit_transform(df['Gender'])
df['City_Encoded'] = le.fit_transform(df['City'])

print("\nAfter Label Encoding:")
print(df)

# 6. Encode categorical variables using One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=['Gender', 'City'], dtype=int)

print("\nAfter One-Hot Encoding:")
print(df_encoded)

# 7. Display the final dataset
print("\nFinal Preprocessed Dataset:")
print(df_encoded)

print("\nData preprocessing completed successfully!")

```
